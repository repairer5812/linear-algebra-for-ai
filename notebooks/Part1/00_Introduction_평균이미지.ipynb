## 이 노트북의 약속

이 노트북 한 개로 다음을 체득한다.

1. **이미지는 벡터다** — $\mathbf{x} \in \mathbb{R}^{784}$
2. **데이터셋은 행렬이다** — $X \in \mathbb{R}^{n \times 784}$
3. **벡터들의 평균은 또 하나의 벡터** — $\mathbf{m} = \frac{1}{n}\sum_i \mathbf{x}_i$
4. **이 한 줄 수학이 NumPy 한 줄**: `X[y==7].mean(axis=0)`
5. **결과는 시각적으로 즉시 보인다** — "7의 평균"이 흐릿한 7 모양

오늘은 정의·증명을 다 알 필요 없다. **"수학 정의 한 줄이 코드와 그림으로 직결된다"는 감각만** 가져가면 충분.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
print('NumPy:', np.__version__)
```

## 1. 이미지를 벡터로

MNIST 손글씨 숫자 — `sklearn`에 내장된 8×8 mini-MNIST를 쓴다 (의존성 없음, 항상 가용).

```python
from sklearn.datasets import load_digits

digits = load_digits()
X = digits.data       # (1797, 64) — 이미 벡터 형태로 펴진 행렬
X_img = digits.images # (1797, 8, 8) — 시각화용
y = digits.target     # (1797,) — 0~9 레이블

print(f'데이터셋:    X.shape = {X.shape}  ← n=1797개 이미지, 각 64차원 벡터')
print(f'시각화용:    X_img.shape = {X_img.shape}  ← 8×8 행렬로 다시 본 형태')
print(f'레이블:      y.shape = {y.shape}, 종류 = {np.unique(y)}')
print(f'데이터 한 장: X[0] = {X[0][:20]}... (앞 20개 픽셀)')
```

**관찰**: 
- 이미지 1장 = 64개 숫자의 묶음 = $\mathbb{R}^{64}$의 **벡터**
- 데이터셋 = 1797개 벡터를 행으로 쌓은 **행렬** $\mathbb{R}^{1797 \times 64}$

이게 "이미지가 행렬이다"의 진짜 의미. 1·2회차에서 정식 다룬다.

```python
# 첫 10장 시각화 — 우리가 평균낼 데이터
fig, axes = plt.subplots(1, 10, figsize=(15, 2))
for i, ax in enumerate(axes):
    ax.imshow(X_img[i], cmap='gray_r')
    ax.set_title(f'label={y[i]}')
    ax.axis('off')
plt.suptitle('데이터 첫 10장 — 사람마다 다른 손글씨')
plt.tight_layout()
plt.show()
```

## 2. 한 줄의 수학 — 벡터 평균

숫자 "7"이라고 적힌 모든 이미지의 집합 $\{\mathbf{x}_i : y_i = 7\}$를 모은 뒤, **벡터 평균**을 구한다.

$$\mathbf{m}_7 = \frac{1}{n_7}\sum_{i:\,y_i=7}\mathbf{x}_i \;\in\; \mathbb{R}^{64}$$

이 한 식의 의미:
- 같은 숫자라고 적힌 모든 손글씨를 **벡터로 더한다** (1회차의 벡터 덧셈)
- $\frac{1}{n_7}$을 곱해 평균 (1회차의 스칼라곱)
- 결과는 또 하나의 **64차원 벡터**

다시 8×8 행렬로 펴면 그림이 된다.

### 한 줄의 코드

```python
mean_seven = X[y == 7].mean(axis=0).reshape(8, 8)

plt.figure(figsize=(4, 4))
plt.imshow(mean_seven, cmap='gray_r')
plt.title(f'"7"의 평균 이미지  (n={(y==7).sum()}개 평균)')
plt.axis('off')
plt.show()
```

**관찰**: 
- 흐릿한 "7" 모양이 보인다
- 한 사람의 7이 아니라 **수많은 7의 공통된 모양**
- 가장자리는 흐림 (사람마다 다르게 쓰는 부분), 중심은 진함 (모두 일관된 부분)

→ **평균이라는 단순 연산이 "7의 본질"을 드러낸다.** 이게 LA가 데이터에 하는 첫 일.

## 3. 0~9의 모든 평균 이미지

10번 반복으로 모든 숫자의 평균을 본다 — for문 한 줄.

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 4.5))
for k in range(10):
    ax = axes[k // 5][k % 5]
    mean_k = X[y == k].mean(axis=0).reshape(8, 8)
    ax.imshow(mean_k, cmap='gray_r')
    ax.set_title(f'"{k}" 평균 (n={(y==k).sum()})')
    ax.axis('off')
plt.suptitle('각 숫자의 평균 이미지 — 흐릿한 0, 1, 2, ..., 9')
plt.tight_layout()
plt.show()
```

**관찰**:

- 0, 1, 2, ..., 9의 평균이 모두 **흐릿하지만 알아볼 수 있는 형태**로 나타남
- "평균"이라는 한 연산이 데이터셋의 본질을 압축
- 1, 7처럼 단순한 모양은 평균이 진하게, 8처럼 복잡한 모양은 흐릿하게

**메시지**: 벡터 평균이라는 1회차 첫 연산만으로도 "클래스의 대표"가 만들어진다.

## 4. 응용 — 가장 단순한 분류기

새 이미지가 들어왔을 때, **가장 가까운 평균**의 레이블로 분류하면 어떨까?

수학:
$$\hat{y} = \arg\min_{k \in \{0, \ldots, 9\}} \|\mathbf{x} - \mathbf{m}_k\|_2$$

여기서 $\|\cdot\|_2$는 $\ell_2$ 노름 (1회차에서 정식 정의). 단순한 분류기지만 의외로 잘 동작.

```python
# 10개 평균 이미지 계산 (행렬 형태로)
means = np.stack([X[y == k].mean(axis=0) for k in range(10)])  # (10, 64)
print(f'평균 이미지 행렬: means.shape = {means.shape}')

# 각 데이터에서 가장 가까운 평균 찾기
# X[i]와 means의 모든 행 사이 유클리드 거리: shape (1797, 10)
dists = np.linalg.norm(X[:, None, :] - means[None, :, :], axis=2)
y_pred = np.argmin(dists, axis=1)

accuracy = (y_pred == y).mean()
print(f'\n분류 정확도: {accuracy*100:.2f}%')
print('→ 평균 한 줄로 만든 분류기인데 80% 넘게 맞춘다')
```

```python
# 어디서 틀리는지 — 혼동행렬
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y, y_pred)

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm, cmap='Blues')
ax.set_xticks(range(10))
ax.set_yticks(range(10))
ax.set_xlabel('예측 (가장 가까운 평균)')
ax.set_ylabel('실제 레이블')
ax.set_title('혼동행렬 — 어디서 틀리나')
for i in range(10):
    for j in range(10):
        ax.text(j, i, str(cm[i, j]), ha='center', va='center',
                color='white' if cm[i, j] > 50 else 'black', fontsize=10)
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()

print('관찰: 대각선이 진하면 잘 맞춤. 헷갈리는 쌍 (예: 4↔9, 8↔3)이 흐릿하게 보임.')
```

## 5. 보너스 — 평균을 빼면 무엇이 보이는가 (PCA 미리보기)

각 "7" 이미지에서 "7의 평균"을 빼면 무엇이 남을까?

$$\mathbf{x}_i - \mathbf{m}_7 \;=\; \text{각 개인의 7이 평균에서 얼마나 벗어났나}$$

이 **편차**들의 분포가 곧 "7의 다양성". 그 다양성의 주된 방향을 찾는 것이 **PCA** (Part 2 5회차).

```python
X_7 = X[y == 7]                  # 7만 모은 행렬
centered = X_7 - X_7.mean(axis=0) # 평균을 뺀 (중심화된) 데이터

# 첫 4개 "7"의 편차 시각화 — 원본 vs 평균 vs 편차
fig, axes = plt.subplots(3, 4, figsize=(10, 7))
for j in range(4):
    axes[0, j].imshow(X_7[j].reshape(8, 8), cmap='gray_r')
    axes[0, j].set_title(f'원본 7 (#{j})')
    axes[0, j].axis('off')
    axes[1, j].imshow(X_7.mean(axis=0).reshape(8, 8), cmap='gray_r')
    axes[1, j].set_title('7의 평균')
    axes[1, j].axis('off')
    axes[2, j].imshow(centered[j].reshape(8, 8), cmap='seismic', vmin=-8, vmax=8)
    axes[2, j].set_title('편차 (원본 - 평균)')
    axes[2, j].axis('off')
plt.tight_layout()
plt.show()

print('편차 이미지의 빨강·파랑 = 평균에서 위·아래로 벗어난 방향. 사람마다 다른 7의 "개성".')
```

**관찰**: 
- 평균은 모든 7의 "중심"
- 편차는 각 7의 "개성" — 어디가 더 진하고 어디가 더 흐린지
- 이 편차들의 **주된 방향**을 찾는 것이 **PCA** (Part 2 5회차) — 차원 축소·임베딩의 출발점

지금은 "평균을 빼는 게 의미가 있다"는 직관만 가져가면 충분.

## 6. 정리 — 이 한 시연 안에 들어 있는 30회차의 지점

| 시연의 부분 | 회차 |
|---|:---:|
| 이미지를 벡터로 ($\mathbb{R}^{64}$) | **1** |
| 벡터 덧셈·스칼라곱 (평균 연산) | **1** |
| 데이터셋을 행렬로 ($\mathbb{R}^{n\times 64}$) | **2** |
| 유클리드 거리 ($\ell_2$) 기반 분류 | **1·10** |
| 평균 중심화 — 데이터의 원점 | **7 · Part 2 8** |
| 편차의 주방향 = PCA | **Part 2 5** |

**가장 단순한 수학(벡터 평균) 한 줄이 30회차 전체의 입구.**
