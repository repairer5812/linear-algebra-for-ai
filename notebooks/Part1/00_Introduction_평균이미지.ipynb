## 이 노트북의 약속

이 노트북 한 개로 다음을 체득한다.

1. **이미지는 벡터다** — $\mathbf{x} \in \mathbb{R}^{784}$
2. **데이터셋은 행렬이다** — $X \in \mathbb{R}^{n \times 784}$
3. **벡터들의 평균은 또 하나의 벡터** — $\mathbf{m} = \frac{1}{n}\sum_i \mathbf{x}_i$
4. **이 한 줄 수학이 NumPy 한 줄**: `X[y==7].mean(axis=0)`
5. **결과는 시각적으로 즉시 보인다** — "7의 평균"이 흐릿한 7 모양

오늘은 정의·증명을 다 알 필요 없다. **"수학 정의 한 줄이 코드와 그림으로 직결된다"는 감각만** 가져가면 충분.

## 1. 이미지를 벡터로

MNIST 손글씨 숫자 — sklearn으로 **28×28 진짜 MNIST**($\mathbb{R}^{784}$)를 내려받는다 (최초 1회 ~15초 다운로드, 이후 캐시).

```python
from sklearn.datasets import fetch_openml

# 28x28 진짜 MNIST 70000장 (데이터 로드용 — 수학이 아니라 다운로드. 최초 1회만 받고 캐시).
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X = mnist.data.astype(float)       # (70000, 784) — 이미 벡터로 펴진 행렬
y = mnist.target.astype(int)       # (70000,) — 0~9 레이블
X_img = X.reshape(-1, 28, 28)      # (70000, 28, 28) — 시각화용

print(f'데이터셋:    X.shape = {X.shape}  ← n=70000개 이미지, 각 784차원 벡터')
print(f'시각화용:    X_img.shape = {X_img.shape}  ← 28×28 행렬로 다시 본 형태')
print(f'레이블:      y.shape = {y.shape}, 종류 = {np.unique(y)}')
print(f'데이터 한 장: X[0] = {X[0][:20]}... (앞 20개 픽셀)')
```

**관찰**:
- 이미지 1장 = 784개 숫자의 묶음 = $\mathbb{R}^{784}$의 **벡터**
- 데이터셋 = 70000개 벡터를 행으로 쌓은 **행렬** $\mathbb{R}^{70000 \times 784}$

이게 "이미지가 행렬이다"의 진짜 의미. 1·2회차에서 정식 다룬다.

## 2. 한 줄의 수학 — 벡터 평균

숫자 "7"이라고 적힌 모든 이미지의 집합 $\{\mathbf{x}_i : y_i = 7\}$를 모은 뒤, **벡터 평균**을 구한다.

$$\mathbf{m}_7 = \frac{1}{n_7}\sum_{i:\,y_i=7}\mathbf{x}_i \;\in\; \mathbb{R}^{784}$$

이 한 식의 의미:
- 같은 숫자라고 적힌 모든 손글씨를 **벡터로 더한다** (1회차의 벡터 덧셈)
- $\frac{1}{n_7}$을 곱해 평균 (1회차의 스칼라곱)
- 결과는 또 하나의 **784차원 벡터**

다시 28×28 행렬로 펴면 그림이 된다.

```python
# .mean(axis=0)이 수학(벡터 평균). .reshape(28,28)은 784벡터를 28×28 격자로 되돌리는
# 표시용 변형이고 cmap='gray_r'도 시각화 옵션 — 둘 다 수학 아님.
mean_seven = X[y == 7].mean(axis=0).reshape(28, 28)

plt.figure(figsize=(4, 4))
plt.imshow(mean_seven, cmap='gray_r')
plt.title(f'"7"의 평균 이미지  (n={(y==7).sum()}개 평균)')
plt.axis('off')
plt.show()
```

**관찰**:
- 또렷한 "7" 모양이 보인다
- 한 사람의 7이 아니라 **수많은 7의 공통된 모양**
- 가장자리는 흐림(사람마다 다른 부분), 중심은 진함(모두 일관된 부분)

→ **평균이라는 단순 연산이 "7의 본질"을 드러낸다.** 이게 LA가 데이터에 하는 첫 일.

> **한 걸음 더**: 이 평균들로 "가장 가까운 평균"을 골라 분류만 해도 80% 안팎을 맞춥니다. 거리·분류의 정식 정의는 2회차에서 다룹니다.

## 3. 0~9의 모든 평균 이미지

10번 반복으로 모든 숫자의 평균을 본다 — for문 한 줄.

```python
# 수학은 각 숫자의 .mean(axis=0) 한 줄. reshape·imshow·subplots는 표시용 시각화 메커닉(수학 아님).
fig, axes = plt.subplots(2, 5, figsize=(10, 4.5))
for k in range(10):
    ax = axes[k // 5][k % 5]
    mean_k = X[y == k].mean(axis=0).reshape(28, 28)
    ax.imshow(mean_k, cmap='gray_r')
    ax.set_title(f'"{k}" 평균 (n={(y==k).sum()})')
    ax.axis('off')
plt.suptitle('각 숫자의 평균 이미지 — 또렷한 0, 1, 2, ..., 9')
plt.tight_layout()
plt.show()
```

**관찰**:

- 0, 1, 2, ..., 9의 평균이 모두 **흐릿하지만 알아볼 수 있는 형태**로 나타남
- "평균"이라는 한 연산이 데이터셋의 본질을 압축
- 1, 7처럼 단순한 모양은 평균이 진하게, 8처럼 복잡한 모양은 흐릿하게

**메시지**: 벡터 평균이라는 1회차 첫 연산만으로도 "클래스의 대표"가 만들어진다.

## 4. 정리 — 이 한 시연 안에 들어 있는 28회차의 지점

| 시연의 부분 | 회차 |
|---|:---:|
| 이미지를 벡터로 ($\mathbb{R}^{784}$) | **1** |

**가장 단순한 수학(벡터 평균) 한 줄이 28회차 전체의 입구.**
