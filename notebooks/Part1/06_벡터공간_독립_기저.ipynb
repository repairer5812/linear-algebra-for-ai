# 6회차 실습: 공간과 그 크기를 **눈으로** (span · 독립 · Null space)

교안과 **같은 벡터** $\mathbf{a}_1=(1,0,1),\ \mathbf{a}_2=(2,1,3),\ \mathbf{a}_3=(0,1,1)$로 세 가지를 본다.

1. **span** = 재료로 만들 수 있는 모든 것 (평면 or 직선)
2. **종속** = 새 벡터가 이미 만든 공간 안에 눕는다
3. **Null space** = 행렬이 원점으로 보내 버리는 방향


```python
!pip install -q koreanize-matplotlib
import numpy as np, matplotlib.pyplot as plt, koreanize_matplotlib  # noqa
from mpl_toolkits.mplot3d import Axes3D  # noqa
CR, NV, GD = '#862633', '#22386B', '#9A6B00'
a1, a2 = np.array([1., 0, 0]), np.array([0., 1, 1])   # 잘 벌어진 두 벡터(90도) -> 평면이 뚜렷
a3 = a2 - 2*a1                                          # = (-2, 1, 1), 평면 안(종속)
A = np.column_stack([a1, a2, a3])
rng = np.random.default_rng(0)
```

## ① span: 재료로 만들 수 있는 모든 것

$\{\mathbf{a}_1,\mathbf{a}_2\}$의 온갖 결합을 찍으면 **평면**을 채운다. 하지만 $\{\mathbf{a}_1, 2\mathbf{a}_1\}$처럼 한 방향뿐이면 **직선**으로 무너진다.


```python
st = rng.uniform(-2.5, 2.5, (2, 800))
P_plane = a1[:, None]*st[0] + a2[:, None]*st[1]        # span{a1,a2} = 평면
P_line  = a1[:, None]*st[0] + (2*a1)[:, None]*st[1]    # span{a1,2a1} = 직선

fig = plt.figure(figsize=(10, 4.4))
for k, (P, ttl) in enumerate([(P_plane, 'span{a1,a2} = 평면 (2차원)'), (P_line, 'span{a1,2a1} = 직선 (1차원)')]):
    ax = fig.add_subplot(1, 2, k+1, projection='3d')
    ax.scatter(P[0], P[1], P[2], s=5, color=CR, alpha=.25)
    ax.set_title(ttl, fontsize=11); ax.view_init(22, -60)
    ax.set_xlim(-3, 3); ax.set_ylim(-3, 3); ax.set_zlim(-3, 3)
    ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
plt.suptitle('span = 재료 벡터로 도달 가능한 모든 점 (같은 축 범위)', fontsize=12)
plt.tight_layout(); plt.show()
```

## ② 종속: $\mathbf{a}_3$는 이미 만든 평면 안에 눕는다

$\mathbf{a}_3 = \mathbf{a}_2 - 2\mathbf{a}_1$이라, $\mathbf{a}_3$은 $\{\mathbf{a}_1,\mathbf{a}_2\}$ 평면을 벗어나지 못한다(**종속**). 반면 $(0,0,1)$은 평면 밖으로 솟는다(**독립**).


```python
g = np.linspace(-2.5, 2.5, 2); S, T = np.meshgrid(g, g)
plane = a1[:, None, None]*S + a2[:, None, None]*T
fig = plt.figure(figsize=(6.4, 5.4)); ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(plane[0], plane[1], plane[2], alpha=.2, color=CR)
for v, c, lab in [(a1, NV, 'a1'), (a2, NV, 'a2'), (a3, GD, 'a3 = a2-2a1 (평면 안)'), (np.array([0.,0,1]), 'k', '(0,0,1) 평면 밖')]:
    ax.quiver(0, 0, 0, *v, color=c, lw=2.6, arrow_length_ratio=.12)
    ax.text(*(v*1.06), lab, color=c, fontsize=10)
ax.set_title('a3는 평면 안(종속), (0,0,1)은 평면 밖(독립)')
ax.view_init(20, -60); ax.set_xlim(-2.5,2.5); ax.set_ylim(-2.5,2.5); ax.set_zlim(-2.5,2.5)
plt.tight_layout(); plt.show()
print('a2 - 2a1 =', a2 - 2*a1, '= a3 ?', np.allclose(a2 - 2*a1, a3))
```

## ③ Null space: 행렬이 원점으로 보내는 방향

$A\mathbf{x}=\mathbf{0}$의 해는 직선 $t\,(2,-1,1)$이다. 이 직선 위의 **어떤 점을 넣어도** $A$는 원점 하나로 보낸다.


```python
n = np.array([2, -1, 1])                     # N(A)의 기저 방향 (정수)
line = n[:, None] * np.linspace(-2, 2, 2)[None, :]
fig = plt.figure(figsize=(6, 5.2)); ax = fig.add_subplot(111, projection='3d')
ax.plot(line[0], line[1], line[2], color=CR, lw=3, label='N(A): (2,-1,1)의 배수')
ax.scatter(0, 0, 0, color='k', s=45); ax.legend(); ax.view_init(18, -58)
ax.set_title('Null space = A가 원점으로 붕괴시키는 직선'); plt.tight_layout(); plt.show()

print('N(A) 위 정수 벡터는 모두 A가 0으로 보낸다:')
for v in [np.array([2, -1, 1]), np.array([4, -2, 2]), np.array([-2, 1, -1])]:
    print(f'  A @ {v} = {A @ v}')
w = np.array([1, 0, 0])
print(f'대조) N(A) 밖 벡터:  A @ {w} = {A @ w}   (0이 아님)')
```

## ④ AI 응용: 데이터를 저차원으로 **압축**

겉보기 3차원 데이터가 실은 **2차원 평면**(span) 위에 있으면, 각 점을 **2개 좌표**로 줄일 수 있다. span·기저·차원이 그대로 압축 도구가 된다.

```python
# 잘 벌어진 평면 기저 (span{u1,u2} = 기울어진 평면, 뚜렷한 2차원)
u1, u2 = np.array([1., 0, 0.6]), np.array([0., 1, 0.6])
B = np.column_stack([u1, u2])          # 부분공간의 기저 (3x2)
C = rng.uniform(-2, 2, (200, 2))       # 각 점의 '평면 좌표' (c1, c2)
X = C @ B.T                            # 3차원 점 = c1*u1 + c2*u2  (평면 위)

# --- 저장에 필요한 숫자 개수 비교 (압축) ---
print(f'점 {len(X)}개, 각 점은 3좌표(x,y,z)  ->  저장에 {len(X)}x3 = {X.size}개 숫자')
print(f'그런데 점이 평면 위 -> 각 점을 평면좌표 2개(c1,c2)로:  {len(C)}x2 = {C.size}개')
print(f'                                    + 평면 기저 2개(3x2=6) = {C.size + 6}개')
X_recon = C @ B.T                      # 2개 좌표로 3차원 복원(재구성)
print(f'재구성 오차(최대): {np.abs(X - X_recon).max():.3f}  ->  {X.size}개를 {C.size+6}개로 손실 없이 압축')
print('* 여기선 평면(기저 u1,u2)을 미리 알고 데이터를 만들었다.')
print('  데이터에서 그 평면을 스스로 찾는 것이 PCA (Part 3).')

# 시각화: 왼쪽 3D(평면 위), 오른쪽 2D(압축된 좌표)
gg = np.linspace(-2, 2, 2); S, T = np.meshgrid(gg, gg)
plane = u1[:, None, None]*S + u2[:, None, None]*T
fig = plt.figure(figsize=(10, 4.2))
ax = fig.add_subplot(1, 2, 1, projection='3d')
ax.plot_surface(plane[0], plane[1], plane[2], alpha=.15, color=CR)
ax.scatter(X[:,0], X[:,1], X[:,2], s=7, color=CR, alpha=.5)
ax.set_title('겉보기 3차원 (하지만 평면 위)'); ax.view_init(22, -60)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
ax2 = fig.add_subplot(1, 2, 2)
ax2.scatter(C[:,0], C[:,1], s=7, color=NV, alpha=.5)
ax2.set_title('실제 2차원: 압축된 좌표 (c1, c2)')
ax2.set_xlabel('c1'); ax2.set_ylabel('c2'); ax2.set_aspect('equal'); ax2.grid(alpha=.3)
plt.suptitle('같은 데이터: 3좌표 -> 2좌표로 압축 (차원 = 2)', fontsize=12)
plt.tight_layout(); plt.show()
```

## 정리

- **span** = 재료로 도달 가능한 전체(평면/직선). **독립** = 낭비 없는 재료. **차원** = 재료 개수.
- **종속 ⟺ Null space ≠ {0}**: $\mathbf{a}_3=\mathbf{a}_2-2\mathbf{a}_1$이 곧 $A(2,-1,1)^\top=\mathbf{0}$.
- 실데이터의 **유효 차원**이 겉보기보다 낮다 → 7회차 rank, Part 2 PCA로 이어진다.
